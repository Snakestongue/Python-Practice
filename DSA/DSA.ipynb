Linked Lists: Made of 2 things, a node (value) and a pointer (pointing to next value).
Unfortunatly manually make it in Python & Java 

```python
class Node:
    def __init__ (self, value):
        self.value = value
        self.next = None

head = Node(1) # first value of linked lists called a head 
curr = head
for i in range(2,11): #for 10 nodes
    curr.next = Node(i)
    curr = curr.next

newCurr = head
while newCurr:
    print(newCurr.value)
    newCurr = newCurr.next
```

BST: Left small Right Big!

```python
class Node:
    def __init__ (self, value):
        self.value = value
        self.left = None
        self.right = None

def insert(current, value): #current is like the head, value is the value i wanna insert
    if current is None:
        return Node(value)
    if value < current.value:
        current.left = insert(current.left, value)
    else:
        current.right = insert(current.right, value)
    return current
```

Heaps: Built in library --> Max Value or Minimum Value on top

```python
import heapq
heap = []
heapq.heappush(heap, 27)
heapq.heappush(heap, 18)
heapq.heappush(heap, 3)

heap2 = [1,2,3,4,5,6,5,4,3,2,1]
heapq.heapify(heap2)
print(heap2)
print(heap)
```
